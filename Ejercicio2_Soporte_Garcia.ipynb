## Ejercicio 2: Escalamiento de tickets de soporte técnico

Agente que decide a qué nivel de soporte enviar un ticket según el tiempo de espera, la urgencia reportada y si el cliente es premium.

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | tiempo_espera_minutos, nivel_urgencia, cliente_premium |
| Acciones (A) | resolver en nivel 1, escalar a nivel 2, escalar a nivel 3 |
| Entorno (E) | Mesa de ayuda con tickets entrantes de clientes distintos, con urgencias y tiempos de espera variables |
| Objetivo | Resolver cada ticket en el nivel adecuado a su urgencia y prioridad |
| Medida de desempeño (P) | Tiempo de resolución acorde a la urgencia, tickets mal escalados, satisfacción del cliente |

### Justificación de las reglas

La urgencia alta se escala de inmediato porque no vale la pena arriesgar tiempo de espera adicional evaluando otras condiciones. Para urgencia media, el tiempo de espera y la condición premium sí importan, porque un ticket que lleva esperando demasiado o que es de un cliente premium merece pasar a un nivel con más capacidad de resolución. La urgencia baja se mantiene en nivel 1 salvo un caso: cliente premium con espera muy larga, porque ahí el compromiso comercial pesa más que la urgencia técnica reportada.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta reportada por el cliente"

    elif nivel_urgencia == "media":
        if tiempo_espera_minutos > 30 or cliente_premium:
            return "escalar a nivel 2", "urgencia media con espera prolongada o cliente premium"
        else:
            return "resolver en nivel 1", "urgencia media con espera corta y cliente no premium"

    else:  # baja
        if cliente_premium and tiempo_espera_minutos > 60:
            return "escalar a nivel 2", "urgencia baja pero cliente premium con espera muy larga"
        else:
            return "resolver en nivel 1", "urgencia baja sin condiciones que ameriten escalamiento"
```
