## Ejercicio 3: Inspección visual de un producto

Agente que revisa una imagen de 5x5 píxeles de un producto (0 = normal, 1 = defectuoso) y decide si aprobarlo, mandarlo a revisión manual o rechazarlo según la cantidad de píxeles defectuosos.

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| Percepción (S) | Matriz 5x5 capturada por la cámara (0 = normal, 1 = defectuoso) |
| Acciones (A) | Aprobar, enviar a revisión manual, rechazar |
| Entorno (E) | Línea de producción, con productos pasando de forma continua frente a la cámara |
| Objetivo | Filtrar productos defectuosos sin frenar innecesariamente la producción |
| Medida de desempeño (P) | Defectos detectados correctamente, productos buenos rechazados por error, productos malos aprobados por error |

### Justificación de las reglas

Cero defectos es el único caso sin ambigüedad, así que se aprueba directo. Entre 1 y 3 píxeles defectuosos puede tratarse de ruido de la cámara o un defecto menor real, y distinguir eso a simple vista no es confiable, por lo que se manda a revisión manual en vez de decidir automáticamente. Más de 3 píxeles ya es un patrón de defecto claro, así que se rechaza sin pasar por revisión.
